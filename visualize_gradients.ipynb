```python
import numpy as np
import dit

import matplotlib.pyplot as plt
```

```python
f1 = np.linspace(450, 550, 200)
f2 = 500

d = dit.cost.harmonic(f1, f2, kernel="berezovsky", gradient=False, berezovsky_erb=False)
dd = dit.cost.harmonic(f1, f2, kernel="berezovsky", gradient=True, berezovsky_erb=False)

p = 112
grad_tangent = dd[p] * (f1 - f1[p]) + d[p]

plt.figure(figsize=(8, 3))

plt.title("Kernel: Berezovsky (2019)")

plt.plot(f1, d, label="kernel")
plt.plot(f1, dd, label="gradient")
plt.plot(f1, grad_tangent, label="tangent")
plt.plot(f1[p], d[p], "x")

plt.ylim(-0.25, 1.05)
plt.ylabel("Cost")

plt.xlabel("Frequency $f_2$ (Hz)")

plt.legend()
plt.show()
```
